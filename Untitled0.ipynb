```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

```python
import pandas as pd
import zipfile

# Upload the ZIP file in Colab
from google.colab import files
uploaded = files.upload()  # This will prompt you to choose a file

# Get the name of the uploaded file
zip_path = list(uploaded.keys())[0]

# Open the ZIP and read the files
with zipfile.ZipFile(zip_path, 'r') as z:
    print("Files in ZIP:", z.namelist())  # Show all files

    # Read the CSV file
    with z.open('marketing_data.csv') as f_csv:
        df_csv = pd.read_csv(f_csv)

    # Read the Excel file
    with z.open('Data Dictionary - Response to marketing campaigns.xlsx') as f_excel:
        df_excel = pd.read_excel(f_excel)

# Check the first few rows
print("CSV DataFrame:")
print(df_csv.head())

print("\nExcel DataFrame:")
print(df_excel.head())
```

```python
df_csv.describe()
df_excel.describe()
```

```python
df_csv['Dt_Customer'] = pd.to_datetime(df_csv['Dt_Customer'], format='%m/%d/%y')
```

```python
df_csv.isnull().sum()
```

```python
# Strip whitespace from column names to ensure accurate referencing
df_csv.columns = df_csv.columns.str.strip()

# Clean 'Income' column: remove '$' and ',' and convert to numeric
df_csv['Income'] = df_csv['Income'].astype(str).str.replace(r'[$,]', '', regex=True).astype(float)

# Fill missing values using grouped means
df_csv['Income'] = df_csv.groupby(['Education','Marital_Status'])['Income'].transform(lambda x: x.fillna(x.mean()))
```

```python
df_csv.isnull().sum()
```

```python
df_csv['Marital_Status'].unique()
```

```python
df_csv['Marital_Status'] = df_csv['Marital_Status'].replace({
'Alone':'Single',
'YOLO':'Single',
'Absurd':'Single'
})
```

```python
df_csv['Total_Children'] = df_csv['Kidhome'] + df_csv['Teenhome']
```

```python
df_csv['Age'] = 2024 - df_csv['Year_Birth']
```

```python
df_csv['Total_Spending'] = (
df_csv['MntWines'] +
df_csv['MntFruits'] +
df_csv['MntMeatProducts'] +
df_csv['MntFishProducts'] +
df_csv['MntSweetProducts'] +
df_csv['MntGoldProds']
)
```

```python
df_csv['Total_Purchases'] = (
df_csv['NumWebPurchases'] +
df_csv['NumCatalogPurchases'] +
df_csv['NumStorePurchases']
)
```

```python
df_csv.hist(figsize=(15,10))
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df_csv['Income'])
plt.show()
```

```python
Q1 = df_csv['Income'].quantile(0.25)
Q3 = df_csv['Income'].quantile(0.75)

IQR = Q3 - Q1

df_csv = df_csv[(df_csv['Income'] >= Q1 - 1.5*IQR) & (df_csv['Income'] <= Q3 + 1.5*IQR)]
```

```python
education_order = {
'Basic':0,
'2n Cycle':1,
'Graduation':2,
'Master':3,
'PhD':4
}

df_csv['Education'] = df_csv['Education'].map(education_order)
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(df_csv.select_dtypes(include=np.number).corr(), cmap='coolwarm')
plt.show()
```
